Read file -> entire, line by line, read all lines as list

```python
f = open("amit.txt", "r")

content = f.read()

print(content)

f.close()
```

```python
with open("amit.txt",'r') as f :
    content = f.read()
    print(content)
    
    
# ✔ automatically closes file
# ✔ safe
# ✔ clean
```

```python
# 4. Reading File Line by Line (IMPORTANT)


with open("amit.txt","r") as f:
    for line in f:
        # print(line)
        print(line.strip())
```

```python
with open("amit.txt","w+") as f:
    f.write("Hello Python \n")
    f.write("Learning File Handling \n")
    f.seek(0)
    print(f.read())
```

```python
f = open("data.txt",'r')
lines = f.readlines()
print(lines)
f.close()
```

```python
f = open("data.txt", "a")
f.write("\nNew Line Added")
f.close()
```

```python
with open("data.txt") as f:
    for line in f:
        print(line)
```

```python
with open("data.txt") as f :
    print(f.tell())
    
    f.seek(0)
    print(f.read())
```
