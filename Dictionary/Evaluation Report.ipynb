```python
from collections import defaultdict


scores = {
    "accuracy": 0.91,
    "precision": 0.88,
    "recall": 0.85
}

score_state = {
    0.9 : "excellent",
    0.85 : "good"
}


def evaluate(scores : dict):
    result = {}
    
    for parameter, score in scores.items():
        
        status = "average"
        
        if score >= 0.9 : score = "excellent"
        elif score >=0.85 : score = "good"       
        cal = {
            "value" : parameter,
            "status" : status
        }     
        result[parameter] = cal
        
    return result


print(evaluate(scores))
        
    
```

```python
scores = {
    "accuracy": 0.91,
    "precision": 0.88,
    "recall": 0.85
}



thresholds = [
    (0.9, "excellent"),
    (0.85, "good"),
    (0, "average")
]

report = {}

for metric , value in scores.items():
    for threshold, label in thresholds:
        if value >=threshold:
            report[metric] = {
                "value" : value,
                "status": label
            }
            break


print(report)


```
