```python
class balance:
    a = 10
    _b = 14
    __c = 20
    def __init__(self,id,bal):
        self.userid = id
        self._balance = bal
        

ac = balance("u1",100)

print(ac.a)
print(ac._b)
print(ac._balance__c)
```

```python
class UserAccount:
    def __init__(self,userid,balance):
        self.user_id = userid
        self._balance = balance
        
    @property
    def balance(self):
        return self._balance
    
    @balance.setter
    def balance(self, value):
        if value<0:
            raise ValueError("Balance can't be negative")
        self._balance = value
        
user2 = UserAccount("u2", 105)
print(user2.balance)
```

```python
class RateLimiter:
    maxReq = 100
    
    def __init__(self, name):
        self._count = 0
        self.name = name
        
    
    @property
    def reqCount(self):
        return self._count
    
    def reqIncrement(self):
        if self._count >= self.maxReq:
            raise ValueError("Too many Requests")
        self._count += 1
        
req1 = RateLimiter("GET")
print(req1.name)
print(req1.reqCount)
req1.reqIncrement()
req1.reqIncrement()
req1.reqIncrement()
print(req1.reqCount)

```
